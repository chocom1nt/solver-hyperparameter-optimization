# 02. Analysis of experiment results

This notebook assumes you have run at least one experiment and produced `data/results/exp1.csv` (or adjust the path).

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

results_path = "../data/results/exp1.csv"
df = pd.read_csv(results_path)
df.head()
```

```python
print("Status counts:")
print(df["status"].value_counts())
```

```python
plt.figure(figsize=(10, 5))
sns.boxplot(data=df, x="solver_name", y="wall_time")
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```
